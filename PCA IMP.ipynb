```python
data = pd.read_csv('105v2.csv', parse_dates=True, index_col='DateTime',encoding="windows-1252") # или 106.csv
data.head(5)
```

```python
data.info()
```

```python
del data['AffinityIndex'] 
del data['SerialNumber'] 
del data['H'] 
del data['WminH'] 
del data['WmaxH'] 
del data['WindDirectionH(Â°)'] 
del data['TdewPoint(Â°C)'] 
del data['WaterVaporElasticity'] 
del data['AbsolutusHumidity'] 
del data['DensityofAir'] 
del data['SpeedofSoundinAir'] 
```

```python
data = data.groupby(pd.Grouper(freq='D')).mean()  # ПЕРИОД
data
```

```python
data.interpolate(method='linear', inplace=True)
```

```python
print(data.shape)
print(data.head())
data.dtypes
```

```python
data = data.drop('Tmean', axis = 1)
```

```python
scaled = scale(data, with_mean=True, with_std=True)
scaled[:10,:]
```

```python
model = PCA()
model.fit(scaled)
transformed = model.transform(scaled)
```

```python
np.round(model.explained_variance_ratio_ * 100,4)
```

```python
np.round(model.components_,4)
```

```python
transformed = np.round(transformed, 5)
transformed[:10,:6]
```

```python
np.round(np.corrcoef(transformed, rowvar=False), 4)
```
